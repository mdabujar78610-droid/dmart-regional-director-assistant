Analysis & Premise Test

```python
import pandas as pd
import numpy as np

# Load Cleaned Data
df = pd.read_csv("data/store_daily_clean.csv")

# Summary Statistics per Store
print("=== SUMMARY STATS PER STORE ===")
grouped = df.groupby("store").agg(
    mean_revenue=("revenue_inr", "mean"),
    mean_footfall=("footfall", "mean"),
    mean_stockout=("stockout_hours", "mean"),
    total_revenue=("revenue_inr", "sum"),
    total_footfall=("footfall", "sum"),
    mean_achievement=("achievement_pct", "mean"),
    poor_days=("performance_band", lambda x: (x == "Poor").sum())
).reset_index()

grouped["revenue_per_footfall"] = grouped["total_revenue"] / grouped["total_footfall"]
print(grouped[["store", "mean_revenue", "mean_footfall", "mean_stockout", "revenue_per_footfall", "mean_achievement", "poor_days"]])

# Premise Test Ranking
print("\n=== PREMISE TEST (Stockout Hours Ranking) ===")
print(grouped.sort_values(by="mean_stockout", ascending=False)[["store", "mean_stockout"]])

# Correlation Analysis
print("\n=== CORRELATIONS ===")
rev_so_all = np.corrcoef(df["revenue_inr"], df["stockout_hours"])[0, 1]
rev_ff_all = np.corrcoef(df["revenue_inr"], df["footfall"])[0, 1]
print(f"All stores Revenue vs Stockout: {rev_so_all:.3f}")
print(f"All stores Revenue vs Footfall: {rev_ff_all:.3f}")

pune_df = df[df["store"] == "Pune"]
rev_so_pune = np.corrcoef(pune_df["revenue_inr"], pune_df["stockout_hours"])[0, 1]
rev_ff_pune = np.corrcoef(pune_df["revenue_inr"], pune_df["footfall"])[0, 1]
print(f"Pune Revenue vs Stockout: {rev_so_pune:.3f}")
print(f"Pune Revenue vs Footfall: {rev_ff_pune:.3f}")
```
